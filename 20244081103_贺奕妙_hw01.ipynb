```python
def say_hi():
    print("Hello, world!")

def goat_pad(s):
    print("goat" + s + "goat")

def print_n(s, n):
    for _ in range(n):
        print(s)

print("--- 测试 say_hi ---")
say_hi()
print("\n--- 测试 goat_pad ---")
goat_pad("abc")
print("\n--- 测试 print_n ---")
print_n("hi", 3)
```

```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return abs(m)  # 加abs是为了处理负数情况，返回正的公约数

print("(2018, 2019) 的 GCD:", gcd(2018, 2019))
print("(1200, 300) 的 GCD:", gcd(1200, 300))
print("(5040, 60) 的 GCD:", gcd(5040, 60))
print("负数测试 (-12, 8) 的 GCD:", gcd(-12, 8))
```

```python
import math

def euler_limit(n):
    return (1 + 1 / n) ** n

def euler_infinite_sum(n):
    total = 0.0
    for k in range(n):
        total += 1 / math.factorial(k)
    return total

def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    k = 0
    while abs(total - true_e) >= epsilon:
        total += 1 / math.factorial(k)
        k += 1
    return total

def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(f"k={k}: {euler_infinite_sum(k)}")

print("euler_limit(1000):", euler_limit(1000))
print("euler_infinite_sum(4):", euler_infinite_sum(4)) # 应该接近 2.667
print("euler_infinite_sum(0):", euler_infinite_sum(0)) # 应该返回 0
print("euler_approx(0.001):", euler_approx(0.001))
print("\n--- 打印 euler 求和表 ---")
print_euler_sum_table(5)
```

```python
def is_power_of_2(n):
    if n <= 0:
        return False
    while n > 1 and n % 2 == 0:
        n //= 2
    return n == 1

def is_power(b, n):
    if b <= 0 or n <= 0:
        return False
    if b == 1:
        return n == 1
    while n > 1 and n % b == 0:
        n //= b
    return n == 1

print("--- 测试 is_power_of_2 ---")
print("1:", is_power_of_2(1))
print("8:", is_power_of_2(8))
print("10:", is_power_of_2(10))

print("\n--- 测试 is_power ---")
print("is_power(2, 8):", is_power(2, 8))
print("is_power(3, 27):", is_power(3, 27))
print("is_power(3, 10):", is_power(3, 10))
```
