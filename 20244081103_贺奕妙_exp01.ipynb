```python
def score_level(score):
    # 检查范围
    if not 0 <= score <= 100:
        raise ValueError("分数必须在0到100之间")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

# 测试函数
for s in [95, 85, 70, 50]:
    print(f"{s}分：{score_level(s)}")

# 测试异常
try:
    score_level(105)
except ValueError as e:
    print("捕获异常：", e)
```

```python
# 1到100求和，for循环
total_for = 0
for i in range(1, 101):
    total_for += i
print("for循环求和:", total_for)

# 1到100求和，while循环
total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1
print("while循环求和:", total_while)

# 递归阶乘
def factorial_rec(n):
    if n < 0:
        raise ValueError("n必须是非负整数")
    if n == 0:
        return 1
    return n * factorial_rec(n - 1)

# 迭代阶乘
def factorial_iter(n):
    if n < 0:
        raise ValueError("n必须是非负整数")
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print("\n阶乘结果对比：")
for n in [0, 1, 5, 10]:
    print(f"n={n}: 递归={factorial_rec(n)}, 迭代={factorial_iter(n)}")
```

```python
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    return min(values), max(values), sum(values) / len(values)

# 元组解包
mn, mx, avg = summary([85, 58, 92, 76, 60, 89])
print(f"最小值：{mn}, 最大值：{mx}, 平均值：{avg:.2f}")

def sum_all(*args):
    return sum(args)

print("\n测试sum_all：")
print("无参数:", sum_all())
print("单参数:", sum_all(5))
print("多参数:", sum_all(1, 2, 3, 4))
```

```python
# 模拟成绩数据
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

# 1. 使用zip构建字典，并逐项输出
score_dict = dict(zip(names, scores))
print("任务1：字典遍历")
for name, score in score_dict.items():
    print(f"{name}: {score}")

# 2. 使用enumerate输出序号（从1开始）
print("\n任务2：带序号的遍历")
for idx, (name, score) in enumerate(score_dict.items(), start=1):
    print(f"序号{idx}: {name}, 成绩{score}")

# 3. 列表推导式：提取及格学生，生成平方列表
passed_names = [name for name, score in score_dict.items() if score >= 60]
score_squares = [score ** 2 for score in scores]
print(f"\n任务3：\n及格学生名单: {passed_names}\n成绩平方列表: {score_squares}")

# 4. 计算统计量
n = len(scores)
avg = sum(scores) / n
max_score = max(scores)
min_score = min(scores)
pass_rate = sum(1 for s in scores if s >= 60) / n

print(f"\n任务4：\n人数: {n}\n平均分: {avg:.2f}\n最高分: {max_score}\n最低分: {min_score}\n及格率: {pass_rate:.2%}")
```

```python
import os

# 1. 查看当前工作目录
print("当前工作目录：", os.getcwd())

# 2. 创建数据目录 exp01_data
data_dir = "exp01_data"
os.makedirs(data_dir, exist_ok=True)

# 3. 定义文件路径
students_path = os.path.join(data_dir, "students.txt")
report_path = os.path.join(data_dir, "report.txt")

# 4. 写入 students.txt
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

with open(students_path, "w", encoding="utf-8") as f:
    f.write(content)

print("\n数据文件已写入：", students_path)
```

```python
def parse_student(line):
    # 去除首尾换行和空格，按逗号分割
    parts = line.strip().split(",")
    
    # 检查字段数量
    if len(parts) != 2:
        raise ValueError("字段数量不是2（可能缺逗号或多余逗号）")
    
    name, score_str = parts
    
    # 检查姓名是否为空
    if not name:
        raise ValueError("姓名为空")
        
    # 尝试转换为整数
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩 '{score_str}' 不是合法的整数")
        
    # 检查成绩范围
    if not 0 <= score <= 100:
        raise ValueError(f"成绩 {score} 不在 0~100 范围内")
        
    return {"name": name, "score": score}
```

```python
valid_records = []
invalid_records = []

with open(students_path, "r", encoding="utf-8") as f:
    next(f) # 跳过表头
    for line_no, line in enumerate(f, start=2): # 从第2行开始计数
        try:
            record = parse_student(line)
            valid_records.append(record)
        except ValueError as e:
            # 记录行号和错误原因，继续处理后续记录
            invalid_records.append((line_no, line.strip(), str(e)))

print("有效记录：")
for r in valid_records:
    print(r)

print("\n无效记录：")
for r in invalid_records:
    print(f"行号 {r[0]} | 内容: {r[1]} | 错误原因: {r[2]}")
```

```python
n_valid = len(valid_records)
n_invalid = len(invalid_records)
avg_score = sum(r["score"] for r in valid_records) / n_valid
pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / n_valid

# 写入 report.txt（使用 w 模式）
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数: {n_valid}\n")
    f.write(f"无效记录数: {n_invalid}\n")
    f.write(f"平均分: {avg_score:.2f}\n")
    f.write(f"及格率: {pass_rate:.2%}\n")

print("第一次使用 'w' 模式写入，会覆盖旧内容。")

# 演示 w 和 a 的区别
with open(report_path, "w", encoding="utf-8") as f:
    f.write("第一次写入，会覆盖旧内容\n")

with open(report_path, "a", encoding="utf-8") as f:
    f.write("第二次追加，不会覆盖旧内容\n")

# 读取确认
with open(report_path, "r", encoding="utf-8") as f:
    print("report.txt 当前内容：")
    print(f.read())
```

```python
# 1. 定义 Student 类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"学生: {self.name}, 成绩: {self.score}, 是否及格: {self.is_passed()}"

# 2. 使用实验4中的 valid_records 创建对象
# 注意：如果你重启了内核，valid_records 可能会消失，需要重新运行实验4的读取代码，
# 或者直接使用下面这行临时数据测试：
# valid_records = [{"name": "Alice", "score": 85}, {"name": "Carol", "score": 92}]

students = [Student(r["name"], r["score"]) for r in valid_records]

print("任务2：输出学生对象")
for stu in students:
    print(stu)
```

```python
# 3. 定义继承自 Student 的 GraduateStudent 类
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        # 使用 super() 初始化父类属性
        super().__init__(name, score)
        self.research_topic = research_topic

    # 重写 __str__ 方法
    def __str__(self):
        return f"研究生: {self.name}, 成绩: {self.score}, 及格: {self.is_passed()}, 研究方向: {self.research_topic}"

# 测试继承类
grad_stu = GraduateStudent("Alice", 85, "机器学习")
print("任务3：测试继承类")
print(grad_stu)
```

```python
import copy

def make_original():
    # 每次调用生成一个新的原始对象，包含嵌套列表
    return {"name": "Alice", "scores": [85, 90]}

print("任务4：对象复制测试")

# 1. 直接赋值
orig1 = make_original()
alias = orig1
alias["scores"].append(100)  # 修改内部列表
print("\n【直接赋值】修改后原对象 orig1:", orig1)  # 你会发现它也变了

# 2. 浅拷贝 (copy.copy)
orig2 = make_original()
shallow = copy.copy(orig2)
shallow["scores"].append(100)  # 修改内部列表
print("【浅拷贝】修改后原对象 orig2:", orig2)  # 你会发现它也变了！
print("           浅拷贝的对象 shallow:", shallow)

# 3. 深拷贝 (copy.deepcopy)
orig3 = make_original()
deep = copy.deepcopy(orig3)
deep["scores"].append(100)  # 修改内部列表
print("【深拷贝】修改后原对象 orig3:", orig3)  # 原对象保持不变！
print("           深拷贝的对象 deep:", deep)
```
